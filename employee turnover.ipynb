```python

# Employee Turnover Analytics
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, roc_curve
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, confusion_matrix, classification_report
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline
try:
    from xgboost import XGBClassifier
    HAS_XGB = True
except Exception:
    HAS_XGB = False

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
file_path = r"C:\Users\user\OneDrive\Desktop\HR_comma_sep.csv"  
df = pd.read_csv(r"C:\Users\user\OneDrive\Desktop\HR_comma_sep.csv")
print("Dataset loaded. Shape:", df.shape)
display(df.head())
df.columns = df.columns.str.strip()
print("\n--- DATA QUALITY CHECKS ---")
print("Columns, dtypes:")
print(df.dtypes)

print("\nMissing values (per column):")
print(df.isna().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nNumeric summary:")
display(df.describe())
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
for col in num_cols:
    if df[col].isna().sum() > 0:
        df[col] = df[col].fillna(df[col].mean())
for col in cat_cols:
    if df[col].isna().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing values after imputation (should be 0):")
print(df.isna().sum())
expected_cols = ['satisfaction_level','last_evaluation','number_project','average_montly_hours','time_spend_company','Work_accident','left','promotion_last_5years','salary']
missing_expected = [c for c in expected_cols if c not in df.columns]
if missing_expected:
    print("\nWARNING: These expected columns are missing from the dataset:", missing_expected)
else:
    print("\nAll expected columns exist.")
if 'salary' in df.columns:
    # Some datasets use 'sales' as department name; keep it as is.
    salary_map = {'low':0, 'medium':1, 'high':2}
    df['salary_level'] = df['salary'].map(salary_map)
    # If mapping produced NaNs due to unexpected values, fill with median
    if df['salary_level'].isna().any():
        df['salary_level'] = df['salary_level'].fillna(df['salary_level'].median())
else:
    df['salary_level'] = 0  # fallback
dept_col = None
for candidate in ['sales','Department','dept','department']:
    if candidate in df.columns:
        dept_col = candidate
        break
if dept_col:
    le_dept = LabelEncoder()
    df['department_enc'] = le_dept.fit_transform(df[dept_col].astype(str))
else:
    df['department_enc'] = 0
feature_cols = ['satisfaction_level','last_evaluation','number_project','average_montly_hours','time_spend_company','Work_accident','promotion_last_5years','salary_level','department_enc']
print("\nFeatures used for modeling:", feature_cols)
plt.figure(figsize=(10,8))
num_df = df[feature_cols + ['left']].copy()
corr = num_df.corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap='RdBu', center=0)
plt.title("Correlation matrix (numerical features including 'left')")
plt.show()
corr_with_left = corr['left'].drop('left').sort_values(key=lambda x: x.abs(), ascending=False)
print("\nFeatures most correlated with 'left' (by absolute correlation):")
print(corr_with_left)
plt.figure(figsize=(6,4))
sns.histplot(df['satisfaction_level'], kde=True, bins=30)
plt.title("Distribution: satisfaction_level")
plt.xlabel("Satisfaction Level")
plt.show()

plt.figure(figsize=(6,4))
sns.histplot(df['last_evaluation'], kde=True, bins=30)
plt.title("Distribution: last_evaluation")
plt.xlabel("Last Evaluation")
plt.show()

plt.figure(figsize=(6,4))
sns.histplot(df['average_montly_hours'], kde=True, bins=30)
plt.title("Distribution: average_montly_hours")
plt.xlabel("Average Monthly Hours")
plt.show()
plt.figure(figsize=(8,5))
sns.countplot(x='number_project', hue='left', data=df)
plt.title("Employee count by number_project (hue=left)")
plt.xlabel("Number of Projects")
plt.ylabel("Count")
plt.legend(title='Left', labels=['Stayed (0)','Left (1)'])
plt.show()
proj_summary = df.groupby(['number_project','left']).size().unstack(fill_value=0)
print("\nCounts by number_project and left (rows = #projects):")
display(proj_summary)
leaver_pct = (proj_summary[1] / (proj_summary[0] + proj_summary[1])) * 100
print("\nPercentage of employees who left, by number of projects:")
display(leaver_pct.sort_values(ascending=False))
leavers = df[df['left'] == 1].copy()
X_leavers = leavers[['satisfaction_level','last_evaluation']].values
scaler_cluster = StandardScaler()
X_leavers_s = scaler_cluster.fit_transform(X_leavers)
sil_scores = {}
inertia = {}
for k in range(2,7):
    kmeans = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10)
    labels = kmeans.fit_predict(X_leavers_s)
    sil = silhouette_score(X_leavers_s, labels)
    sil_scores[k] = sil
    inertia[k] = kmeans.inertia_

print("\nSilhouette scores for KMeans (k=2..6):", sil_scores)
print("Inertia (elbow):", inertia)

best_k = max(sil_scores, key=sil_scores.get)
print("Selected k (by silhouette):", best_k)

kmeans = KMeans(n_clusters=best_k, random_state=RANDOM_STATE, n_init=10)
leavers['cluster'] = kmeans.fit_predict(X_leavers_s)
plt.figure(figsize=(7,6))
palette = sns.color_palette("tab10", best_k)
sns.scatterplot(x='satisfaction_level', y='last_evaluation', hue='cluster', data=leavers, palette=palette, s=40)
plt.title(f"KMeans clusters of employees who left (k={best_k})")
plt.xlabel("Satisfaction Level")
plt.ylabel("Last Evaluation")
plt.legend(title='Cluster')
plt.show()
print("\nCluster sizes (among leavers):")
print(leavers['cluster'].value_counts().sort_index())

print("\nCluster centers (scaled space -> original space):")
centers_scaled = kmeans.cluster_centers_
centers_orig = scaler_cluster.inverse_transform(centers_scaled)
centers_df = pd.DataFrame(centers_orig, columns=['satisfaction_level','last_evaluation'])
display(centers_df)
X = df[feature_cols].copy()
y = df['left'].astype(int).copy()
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
class_counts = y.value_counts()
print("\nClass distribution (0=stay,1=left):")
print(class_counts)
print("Left ratio: {:.3f}".format(class_counts[1] / class_counts.sum()))
models = {
    'LogisticRegression': LogisticRegression(random_state=RANDOM_STATE, max_iter=1000),
    'RandomForest': RandomForestClassifier(random_state=RANDOM_STATE, n_estimators=200)
}
if HAS_XGB:
    models['XGBoost'] = XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=RANDOM_STATE)

print("\nModels to evaluate:", list(models.keys()))
skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=RANDOM_STATE)

def evaluate_model_with_smote(estimator, X, y, skf_obj):
    metrics = {'accuracy':[], 'precision':[], 'recall':[], 'f1':[], 'roc_auc':[]}
    fold = 1
    for train_idx, test_idx in skf_obj.split(X, y):
        X_train, X_test = X[train_idx], X[test_idx]
        y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

      
        sm = SMOTE(random_state=RANDOM_STATE)
        X_res, y_res = sm.fit_resample(X_train, y_train)

        
        estimator.fit(X_res, y_res)

       
        y_pred = estimator.predict(X_test)
        
        if hasattr(estimator, "predict_proba"):
            y_proba = estimator.predict_proba(X_test)[:,1]
            roc = roc_auc_score(y_test, y_proba)
        elif hasattr(estimator, "decision_function"):
            y_proba = estimator.decision_function(X_test)
            roc = roc_auc_score(y_test, y_proba)
        else:
            roc = np.nan

       
        metrics['accuracy'].append(accuracy_score(y_test, y_pred))
        metrics['precision'].append(precision_score(y_test, y_pred, zero_division=0))
        metrics['recall'].append(recall_score(y_test, y_pred, zero_division=0))
        metrics['f1'].append(f1_score(y_test, y_pred, zero_division=0))
        metrics['roc_auc'].append(roc)
        fold += 1

    
    summary = {k: (np.nanmean(v), np.nanstd(v)) for k,v in metrics.items()}
    return summary, metrics


results_summary = {}
results_full = {}
for name, model in models.items():
    print(f"\nEvaluating model: {name}")
    model_clone = model
    summary, metrics = evaluate_model_with_smote(model_clone, X_scaled, y, skf)
    results_summary[name] = summary
    results_full[name] = metrics
    print("Mean metrics (mean ± std):")
    for m, (meanv, stdv) in summary.items():
        print(f" {m}: {meanv:.4f} ± {stdv:.4f}")



ranking = []
for name, summ in results_summary.items():
    
    recall_mean = summ['recall'][0]
    f1_mean = summ['f1'][0]
    roc_mean = summ['roc_auc'][0] if not np.isnan(summ['roc_auc'][0]) else 0
    score = (2*recall_mean + f1_mean + roc_mean)  # weighted heuristic
    ranking.append((name, score, recall_mean, f1_mean, roc_mean))
ranking.sort(key=lambda x: x[1], reverse=True)
print("\nModel ranking (heuristic score; higher is better):")
for r in ranking:
    print(f" {r[0]} - score: {r[1]:.4f}, recall: {r[2]:.4f}, f1: {r[3]:.4f}, roc_auc: {r[4]:.4f}")

best_model_name = ranking[0][0]
print("\nSelected best model based on heuristic:", best_model_name)
best_model = models[best_model_name]
sm = SMOTE(random_state=RANDOM_STATE)
X_res_full, y_res_full = sm.fit_resample(X_scaled, y)
best_model.fit(X_res_full, y_res_full)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, stratify=y, test_size=0.2, random_state=RANDOM_STATE)
X_tr_res, y_tr_res = SMOTE(random_state=RANDOM_STATE).fit_resample(X_train, y_train)
best_model.fit(X_tr_res, y_tr_res)
y_pred = best_model.predict(X_test)
if hasattr(best_model, "predict_proba"):
    y_proba = best_model.predict_proba(X_test)[:,1]
    roc = roc_auc_score(y_test, y_proba)
else:
    y_proba = None
    roc = np.nan

print("\nFinal evaluation on holdout test set:")
print(classification_report(y_test, y_pred, digits=4))
print("ROC-AUC:", roc)
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title(f"Confusion Matrix: {best_model_name}")
plt.show()
if y_proba is not None:
    fpr, tpr, _ = roc_curve(y_test, y_proba)
    plt.figure(figsize=(6,5))
    plt.plot(fpr, tpr, label=f'{best_model_name} (AUC={roc:.3f})')
    plt.plot([0,1],[0,1],'--', color='gray')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('ROC Curve')
    plt.legend()
    plt.show()
print("\n--- Feature importance / effect estimates ---")
if best_model_name == 'LogisticRegression':
    coefs = best_model.coef_[0]
    feat_imp = pd.Series(coefs, index=feature_cols).sort_values(key=lambda x: x.abs(), ascending=False)
    display(feat_imp)
else:
    
    try:
        importances = best_model.feature_importances_
        feat_imp = pd.Series(importances, index=feature_cols).sort_values(ascending=False)
        display(feat_imp)
    except Exception:
        print("Feature importance not available for this model.")
print("\n--- Retention Strategies and Recommendations ---\n")
print("1) Prioritize outreach to employees identified as high-risk (model predictions).")
print("   - Use personalized check-ins, improved manager support, and tailored career conversations.")
print("2) Focus on factors that correlate strongly with leaving (from correlation results above).")
print("   - For example, if low satisfaction_level is strongly associated with leaving, deploy targeted satisfaction improvement programs.")
print("3) Improve workload management where high project count or long monthly hours correlate with leaving.")
print("   - Consider rebalancing projects, hiring, or temporary support during peaks.")
print("4) Review compensation packages for low-salary groups and consider adjusted incentives/benefits.")
print("5) Promote internal mobility & clear career paths; those with few promotions might feel stagnant.")
print("6) Implement stay interviews for employees in clusters (from clustering) showing risky satisfaction/evaluation combos.")
print("7) Monitor & re-evaluate model periodically (quarterly) to keep predictions aligned with business reality.")
print("\nOperational suggestions:")
print(" - Integrate model outputs with HR dashboards and alert systems.")
print(" - Design A/B tests for retention interventions to measure impact before broad rollouts.")
print("\nEnd of analysis. For reproducibility, note random_state =", RANDOM_STATE)
```
